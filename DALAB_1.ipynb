```python
a=input("Enter a list of numbers separated by commas: ")
b=a.split(",")
#numbers=[float(num) for num in b]
numbers=list(map(float,b))
#numbers=list(c)
print("Output List: ",numbers)
```

```python
max(numbers)
```

```python
min(numbers)
```

```python
sum(numbers)
```

```python
#round_num=(round(sum(numbers)/len(numbers),2))
round_num=list(map(round, numbers))
print("Rounded Numbers: ",(round_num))
```

```python
average=sum(numbers)/len(numbers)
print("Average of all numbers: ",average)
```
